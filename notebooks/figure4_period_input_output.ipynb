# Figure 4: Period Input-Output Functions

This notebook illustrates how Figure 4 from the manuscript was obtained. For each subfigure, the value of $K_d$ is fixed at the appropriate value, boundaries of the periodic interval are determined, the period is estimated from trajectories, and IPH points are estimated using numerical derivatives. In addition to plotting the graph of the period input-output function, the code cells print the estimated IPH points to compare with the graph and confirm accuracy. 

```python
# import modules
import matplotlib.pyplot as plt
from kim_forger_snf.iph import find_iph_points
```

## Subfigure 4a

The next code cell reproduces Subfigure 4a from the manuscript. After fixing $K_d = 0.0001$, the first step is using `find_iph_points()` to compute boundary points of the periodic interval, estimate the period of the solution over the interval, and return `(A_grid, period, IPH_points)`. The second step uses `A_grid` and `period` to plot the corresponding period input-output function. `IPH_points` are printed to compare with the plot and confirm accuracy. 

```python
# generate data for plot
K_d = 1e-4
A_grid, period, iph_points = find_iph_points(K_d)
print(iph_points)

# create plot
plt.figure(figsize = (12, 8))
plt.plot(A_grid, period, 'r-', linewidth = 2)
plt.xlabel('$A$', fontsize = 24)
plt.xticks(fontsize = 20)
plt.ylabel('Period', fontsize = 24)
plt.yticks(fontsize = 20)
plt.title('Period Input-Output Function with $K_d =$' + f'{K_d}', fontsize = 24)
plt.savefig("../outputs/figure4a.png", dpi = 300, bbox_inches = 'tight')
plt.show()
```

## Subfigure 4b

The next code cell sets $K_d = 0.0015$, then repeats the procedure used above. The result is the plot shown in Subfigure 4b of the manuscript. 

```python
# generate data for plot
K_d = 1.5e-4
A_grid, period, iph_points = find_iph_points(K_d)
print(iph_points)

# create plot
plt.figure(figsize = (12, 8))
plt.plot(A_grid, period, 'r-', linewidth = 2)
plt.xlabel('$A$', fontsize = 24)
plt.xticks(fontsize = 20)
plt.ylabel('Period', fontsize = 24)
plt.yticks(fontsize = 20)
plt.title('Period Input-Output Function with $K_d =$' + f'{K_d}', fontsize = 24)
plt.savefig("../outputs/figure4b.png", dpi = 300, bbox_inches = 'tight')
plt.show()
```
